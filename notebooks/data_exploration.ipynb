## i will remove the outliers as they in negative data and i hope this fix some of imbalancing and not loose information

```python
def remove_outliers(df, columns=None, target='Class', majority_class=0):
    """
    Remove IQR outliers only from the majority class.

    Parameters
    ----------
    df : pd.DataFrame
        Input dataframe.
    columns : list, optional
        Columns to check for outliers. If None, all numeric columns except target are used.
    target : str
        Target column name.
    majority_class : int
        Class from which outliers will be removed.

    Returns
    -------
    pd.DataFrame
        DataFrame with outliers removed from the majority class only.
    """

    df = df.copy()

    if columns is None:
        columns = df.select_dtypes(include='number').columns.drop(target)

    majority = df[df[target] == majority_class].copy()
    minority = df[df[target] != majority_class].copy()

    mask = pd.Series(True, index=majority.index)

    for col in columns:
        Q1 = majority[col].quantile(0.25)
        Q3 = majority[col].quantile(0.75)
        IQR = Q3 - Q1

        lower = Q1 - 1.5 * IQR
        upper = Q3 + 1.5 * IQR

        mask &= majority[col].between(lower, upper)

    majority = majority[mask]

    return pd.concat([majority, minority]).sort_index()
```
